```python
class School():
    def __init__(self,name):
        self.name = name

    def __str__(self):
        return (f'I am the super class: {self.name2}')


class student(School):

    def __init__(self,name1, name2, age):
        self.name2 = name2
        self.age = age
        super().__init__(name1)

```
